# Clean the House Sales Dataset

Download the King County house-sales dataset with KaggleHub, create a few useful features, and save the cleaned file in `Cleaned_Datasets/` for the next notebooks.

## 1. Load and Check the Data

```python
from pathlib import Path

import kagglehub
import pandas as pd
```

```python
dataset_folder = Path(kagglehub.dataset_download("harlfoxem/housesalesprediction"))
data = pd.read_csv(dataset_folder / "kc_house_data.csv")

print(f"Rows and columns: {data.shape}")
print(f"Duplicate rows: {data.duplicated().sum()}")
print(f"Missing values: {data.isna().sum().sum()}")
display(data.head())
```

```python
# Remove exact duplicate rows and add simple features for the sale.
data = data.drop_duplicates().copy()
sale_date = pd.to_datetime(data["date"], format="%Y%m%dT%H%M%S")
data["house_age"] = sale_date.dt.year - data["yr_built"]
data["sale_month"] = sale_date.dt.month
data["renovated"] = (data["yr_renovated"] > 0).astype(int)

# Remove identifiers and source columns replaced by the new features.
data = data.drop(columns=["id", "date", "yr_built", "yr_renovated"])
data = data.dropna(subset=["price"])

cleaned_folder = Path("Cleaned_Datasets")
cleaned_folder.mkdir(exist_ok=True)
data.to_csv(cleaned_folder / "house_sales_clean.csv", index=False)

print(f"Cleaned rows and columns: {data.shape}")
print("Saved: Cleaned_Datasets/house_sales_clean.csv")
```
