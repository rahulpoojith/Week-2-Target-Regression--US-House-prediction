# Random Forest House Price Model

This separate notebook trains a Random Forest model to predict house sale price in USD. It uses the same cleaned data, feature set, and test split as `Model_Training.ipynb` so the scores are comparable.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv("Cleaned_Datasets/house_sales_clean.csv")
features = data.drop(columns=["price", "sqft_above"])
target = data["price"]

X_train, X_test, y_train, y_test = train_test_split(
    features,
    target,
    test_size=0.25,
    random_state=42,
)
print(f"Training rows: {len(X_train)}")
print(f"Test rows: {len(X_test)}")
```

```python
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print("Random Forest model trained.")
```

```python
mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print(f"MAE: ${mae:,.2f}")
print(f"RMSE: ${rmse:,.2f}")
print(f"R-squared: {r2:.3f}")

plt.scatter(y_test, predictions)
plt.show()
```
